```python
# Phase 13 - Reconstruct dynamic graph snapshots
# using the original FinGraph-Sentinel methodology

GRAPH_THRESHOLD = 0.70

companies = [
    col for col in returns_df.columns
    if col != "date"
]

print("Companies:", companies)
print("Number of companies:", len(companies))
print("Correlation threshold:", GRAPH_THRESHOLD)
```

```python
def build_dynamic_graph(returns_window, threshold=0.70):
    """
    Build a dynamic financial correlation graph using
    the original FinGraph-Sentinel methodology.

    Nodes:
        All companies

    Edges:
        Added when absolute correlation >= threshold

    Edge weight:
        Absolute correlation

    Edge attribute:
        Original signed correlation
    """

    correlation_matrix = returns_window.corr()

    G = nx.Graph()

    # Add all companies, including isolated nodes
    for company in companies:
        G.add_node(company)

    # Add edges using the original methodology
    for i in range(len(companies)):
        for j in range(i + 1, len(companies)):

            company_a = companies[i]
            company_b = companies[j]

            correlation = correlation_matrix.loc[
                company_a,
                company_b
            ]

            if pd.isna(correlation):
                continue

            if abs(correlation) >= threshold:

                G.add_edge(
                    company_a,
                    company_b,
                    weight=abs(correlation),
                    correlation=correlation
                )

    return G
```

```python
# ============================================================
# PHASE 13 — RECONSTRUCT EXACT PHASE 8 DYNAMIC GRAPHS
# ============================================================

WINDOW_SIZE = 60
STEP_SIZE = 20
CORRELATION_THRESHOLD = 0.5

graph_snapshots_13 = []

for start in range(
    0,
    len(returns_df) - WINDOW_SIZE + 1,
    STEP_SIZE
):
    end = start + WINDOW_SIZE

    # Exact Phase 8 rolling window
    window = returns_df.iloc[start:end]

    # Correlation matrix
    corr_matrix = window.corr()

    # Create graph
    graph = nx.Graph()

    # Add all companies
    for company in corr_matrix.columns:
        graph.add_node(company)

    # Add correlation edges
    for i in range(len(corr_matrix.columns)):
        for j in range(i + 1, len(corr_matrix.columns)):

            company_a = corr_matrix.columns[i]
            company_b = corr_matrix.columns[j]

            correlation = corr_matrix.iloc[i, j]

            if pd.notna(correlation) and abs(correlation) >= CORRELATION_THRESHOLD:

                distance = 1 - abs(correlation)

                graph.add_edge(
                    company_a,
                    company_b,
                    weight=abs(correlation),
                    distance=distance,
                    correlation=correlation
                )

    snapshot_date = window.index[-1]

    graph_snapshots_13.append({
        "date": snapshot_date,
        "graph": graph,
        "correlation_matrix": corr_matrix
    })

print("Phase 13 graph reconstruction complete.")
print("Number of graph snapshots:", len(graph_snapshots_13))
```

```python
# ============================================================
# RECONSTRUCT PHASE 8 DYNAMIC GRAPHS
# ============================================================

WINDOW_SIZE = 60
STEP_SIZE = 20
CORRELATION_THRESHOLD = 0.5

graph_snapshots_13 = []

for start in range(
    0,
    len(returns_phase13) - WINDOW_SIZE + 1,
    STEP_SIZE
):

    end = start + WINDOW_SIZE

    # Exact 60-observation rolling window
    window = returns_phase13.iloc[start:end]

    # Correlation matrix
    corr_matrix = window.corr()

    # Create graph
    graph = nx.Graph()

    # Add exactly 10 companies
    for company in corr_matrix.columns:
        graph.add_node(company)

    # Add correlation edges
    for i in range(len(corr_matrix.columns)):

        for j in range(i + 1, len(corr_matrix.columns)):

            company_a = corr_matrix.columns[i]
            company_b = corr_matrix.columns[j]

            correlation = corr_matrix.iloc[i, j]

            if (
                pd.notna(correlation)
                and abs(correlation) >= CORRELATION_THRESHOLD
            ):

                distance = 1 - abs(correlation)

                graph.add_edge(
                    company_a,
                    company_b,
                    weight=abs(correlation),
                    distance=distance,
                    correlation=correlation
                )

    # Last date in the rolling window
    snapshot_date = window.index[-1]

    graph_snapshots_13.append({
        "date": snapshot_date,
        "graph": graph,
        "correlation_matrix": corr_matrix
    })

print("Phase 13 graph reconstruction complete.")
print("Number of graph snapshots:", len(graph_snapshots_13))
```

```python
# ============================================================
# PHASE 13 — RELOAD ALL REQUIRED DATA
# ============================================================

import os
import numpy as np
import pandas as pd
import networkx as nx
import torch

from torch_geometric.data import Data

# ------------------------------------------------------------
# Project paths
# ------------------------------------------------------------

PROJECT_ROOT = os.path.abspath("..")

PROCESSED_DIR = os.path.join(
    PROJECT_ROOT,
    "data",
    "processed"
)

RESULTS_DIR = os.path.join(
    PROJECT_ROOT,
    "data",
    "results"
)

print("Project root:", PROJECT_ROOT)
print("Processed:", PROCESSED_DIR)
print("Results:", RESULTS_DIR)


# ------------------------------------------------------------
# Load required datasets
# ------------------------------------------------------------

returns_df = pd.read_csv(
    os.path.join(
        PROCESSED_DIR,
        "returns.csv"
    )
)

systemic_risk_df = pd.read_csv(
    os.path.join(
        RESULTS_DIR,
        "dynamic_systemic_risk.csv"
    )
)

node_evolution_df = pd.read_csv(
    os.path.join(
        RESULTS_DIR,
        "node_network_evolution.csv"
    )
)

ews_df = pd.read_csv(
    os.path.join(
        RESULTS_DIR,
        "phase10_early_warning_dataset.csv"
    )
)


# ------------------------------------------------------------
# Convert dates
# ------------------------------------------------------------

returns_df["date"] = pd.to_datetime(
    returns_df["date"]
)

systemic_risk_df["date"] = pd.to_datetime(
    systemic_risk_df["date"]
)

node_evolution_df["date"] = pd.to_datetime(
    node_evolution_df["date"]
)

ews_df["date"] = pd.to_datetime(
    ews_df["date"]
)


# ------------------------------------------------------------
# Validation
# ------------------------------------------------------------

print("\nDatasets loaded successfully.")

print("\nReturns:")
print(returns_df.shape)

print("\nSystemic risk:")
print(systemic_risk_df.shape)

print("\nNode evolution:")
print(node_evolution_df.shape)

print("\nEWS:")
print(ews_df.shape)
```

```python
# ============================================================
# PHASE 13 — BUILD GRAPH SNAPSHOTS + PyG DATASET
# ============================================================

# ------------------------------------------------------------
# 1. Company configuration
# ------------------------------------------------------------

companies = [
    "AAPL", "AMZN", "GOOGL", "JPM", "MA",
    "META", "MSFT", "NVDA", "TSLA", "V"
]

company_to_idx = {
    company: idx
    for idx, company in enumerate(companies)
}

print("Companies:", companies)


# ------------------------------------------------------------
# 2. Prepare returns
# ------------------------------------------------------------

returns_phase13 = returns_df.copy()

returns_phase13["date"] = pd.to_datetime(
    returns_phase13["date"]
)

returns_phase13 = (
    returns_phase13
    .set_index("date")
    .sort_index()
)

returns_phase13 = returns_phase13[companies]

print("\nReturns shape:", returns_phase13.shape)


# ------------------------------------------------------------
# 3. Reconstruct dynamic graph snapshots
# ------------------------------------------------------------

WINDOW_SIZE = 60
STEP_SIZE = 20
CORRELATION_THRESHOLD = 0.5

graph_snapshots_13 = []

for start in range(
    0,
    len(returns_phase13) - WINDOW_SIZE + 1,
    STEP_SIZE
):

    end = start + WINDOW_SIZE

    window = returns_phase13.iloc[start:end]

    corr_matrix = window.corr()

    G = nx.Graph()

    # Add all companies as nodes
    for company in companies:
        G.add_node(company)

    # Add correlation-based edges
    for i in range(len(companies)):

        for j in range(i + 1, len(companies)):

            company_a = companies[i]
            company_b = companies[j]

            correlation = corr_matrix.loc[
                company_a,
                company_b
            ]

            if (
                pd.notna(correlation)
                and abs(correlation) >= CORRELATION_THRESHOLD
            ):

                G.add_edge(
                    company_a,
                    company_b,
                    weight=float(abs(correlation)),
                    distance=float(1 - abs(correlation)),
                    correlation=float(correlation)
                )

    snapshot_date = window.index[-1]

    graph_snapshots_13.append({
        "date": snapshot_date,
        "graph": G,
        "correlation_matrix": corr_matrix
    })


print("\nGraph snapshots:", len(graph_snapshots_13))


# ------------------------------------------------------------
# 4. Create graph lookup
# ------------------------------------------------------------

graph_lookup = {
    pd.to_datetime(snapshot["date"]): snapshot
    for snapshot in graph_snapshots_13
}


# ------------------------------------------------------------
# 5. Prepare node evolution
# ------------------------------------------------------------

node_features_df = node_evolution_df.copy()

node_features_df["date"] = pd.to_datetime(
    node_features_df["date"]
)

node_features_df = node_features_df[
    node_features_df["ticker"].isin(companies)
].copy()


# ------------------------------------------------------------
# 6. Prepare systemic risk
# ------------------------------------------------------------

systemic_features_df = systemic_risk_df.copy()

systemic_features_df["date"] = pd.to_datetime(
    systemic_features_df["date"]
)

systemic_features_df = systemic_features_df[
    systemic_features_df["company"].isin(companies)
].copy()


# ------------------------------------------------------------
# 7. Prepare EWS target
# ------------------------------------------------------------

ews_target_df = ews_df.copy()

ews_target_df["date"] = pd.to_datetime(
    ews_target_df["date"]
)

ews_target_df["target"] = (
    ews_target_df["future_stress_event"]
    .astype(int)
)


# ------------------------------------------------------------
# 8. Find dates common to all datasets
# ------------------------------------------------------------

graph_dates = set(graph_lookup.keys())

node_dates = set(
    node_features_df["date"].unique()
)

systemic_dates = set(
    systemic_features_df["date"].unique()
)

ews_dates = set(
    ews_target_df["date"].unique()
)

aligned_dates = sorted(
    graph_dates
    & node_dates
    & systemic_dates
    & ews_dates
)

print("Aligned dates:", len(aligned_dates))

print(
    "First aligned date:",
    aligned_dates[0]
)

print(
    "Last aligned date:",
    aligned_dates[-1]
)


# ------------------------------------------------------------
# 9. Build node feature matrices
# ------------------------------------------------------------

node_feature_columns = [
    "degree",
    "strength",
    "betweenness",
    "closeness",
    "eigenvector"
]

node_feature_snapshots = []

for date in aligned_dates:

    node_data = (
        node_features_df[
            node_features_df["date"] == date
        ]
        .set_index("ticker")
    )

    systemic_data = (
        systemic_features_df[
            systemic_features_df["date"] == date
        ]
        .set_index("company")
    )

    feature_rows = []

    for company in companies:

        network_values = node_data.loc[
            company,
            node_feature_columns
        ].values.astype(float)

        systemic_value = float(
            systemic_data.loc[
                company,
                "systemic_risk_score"
            ]
        )

        feature_rows.append(
            np.concatenate([
                network_values,
                [systemic_value]
            ])
        )

    X_nodes = np.vstack(feature_rows)

    target = int(
        ews_target_df.loc[
            ews_target_df["date"] == date,
            "target"
        ].iloc[0]
    )

    node_feature_snapshots.append({
        "date": date,
        "X": X_nodes,
        "target": target
    })


print(
    "\nNode feature snapshots:",
    len(node_feature_snapshots)
)

print(
    "Feature shape:",
    node_feature_snapshots[0]["X"].shape
)


# ------------------------------------------------------------
# 10. Convert to PyTorch Geometric
# ------------------------------------------------------------

pyg_graphs = []

for snapshot in node_feature_snapshots:

    date = pd.to_datetime(snapshot["date"])

    X = snapshot["X"]

    target = snapshot["target"]

    G = graph_lookup[date]["graph"]

    # Node features
    x = torch.tensor(
        X,
        dtype=torch.float32
    )

    # Graph edges
    edge_list = []
    edge_weights = []

    for u, v, attributes in G.edges(data=True):

        u_idx = company_to_idx[u]
        v_idx = company_to_idx[v]

        weight = float(
            attributes["weight"]
        )

        # Undirected graph → both directions
        edge_list.append(
            [u_idx, v_idx]
        )

        edge_list.append(
            [v_idx, u_idx]
        )

        edge_weights.append(weight)
        edge_weights.append(weight)

    if len(edge_list) > 0:

        edge_index = torch.tensor(
            edge_list,
            dtype=torch.long
        ).t().contiguous()

        edge_weight = torch.tensor(
            edge_weights,
            dtype=torch.float32
        )

    else:

        edge_index = torch.empty(
            (2, 0),
            dtype=torch.long
        )

        edge_weight = torch.empty(
            (0,),
            dtype=torch.float32
        )

    # Graph target
    y = torch.tensor(
        [target],
        dtype=torch.float32
    )

    # Create PyG Data object
    data = Data(
        x=x,
        edge_index=edge_index,
        edge_weight=edge_weight,
        y=y
    )

    data.date = date

    pyg_graphs.append(data)


# ------------------------------------------------------------
# 11. Validation
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("PHASE 13 PyG VALIDATION")
print("=" * 70)

print(
    "Graph snapshots:",
    len(graph_snapshots_13)
)

print(
    "Aligned dates:",
    len(aligned_dates)
)

print(
    "PyG graphs:",
    len(pyg_graphs)
)

assert len(graph_snapshots_13) == 60
assert len(aligned_dates) == 21
assert len(pyg_graphs) == 21

for data in pyg_graphs:

    assert data.x.shape == (10, 6)

    assert data.edge_index.shape[0] == 2

    assert (
        data.edge_index.shape[1]
        == data.edge_weight.shape[0]
    )

    assert data.y.shape == (1,)

    assert not torch.isnan(data.x).any()

    assert not torch.isinf(data.x).any()

    if data.edge_index.numel() > 0:

        assert data.edge_index.min() >= 0

        assert data.edge_index.max() < 10


# ------------------------------------------------------------
# 12. Final summary
# ------------------------------------------------------------

targets = [
    int(graph.y.item())
    for graph in pyg_graphs
]

print("\nALL CHECKS PASSED")

print("\nFirst PyG graph:")
print(pyg_graphs[0])

print("\nNode feature shape:")
print(pyg_graphs[0].x.shape)

print("\nEdge index shape:")
print(pyg_graphs[0].edge_index.shape)

print("\nTarget distribution:")
print(
    pd.Series(targets)
    .value_counts()
    .sort_index()
)
```

```python
# ============================================================
# PHASE 13 — TEMPORAL SEQUENCE CONSTRUCTION
# ============================================================

SEQUENCE_LENGTH = 3

temporal_sequences = []

for i in range(
    SEQUENCE_LENGTH,
    len(pyg_graphs)
):

    sequence = pyg_graphs[
        i - SEQUENCE_LENGTH:i
    ]

    target_graph = pyg_graphs[i]

    temporal_sequences.append({
        "graphs": sequence,
        "target": target_graph.y.item(),
        "target_date": target_graph.date
    })


# ------------------------------------------------------------
# Validation
# ------------------------------------------------------------

print("=" * 70)
print("PHASE 13 TEMPORAL SEQUENCE VALIDATION")
print("=" * 70)

print("\nTotal PyG graphs:", len(pyg_graphs))

print(
    "Sequence length:",
    SEQUENCE_LENGTH
)

print(
    "Temporal sequences:",
    len(temporal_sequences)
)

print(
    "First sequence:"
)

for graph in temporal_sequences[0]["graphs"]:
    print(
        "  ",
        graph.date,
        "Target:",
        graph.y.item()
    )

print(
    "  → Prediction target:",
    temporal_sequences[0]["target_date"],
    "Target:",
    temporal_sequences[0]["target"]
)


# ------------------------------------------------------------
# Chronological validation
# ------------------------------------------------------------

for sequence in temporal_sequences:

    dates = [
        graph.date
        for graph in sequence["graphs"]
    ]

    assert dates == sorted(dates)

    assert (
        sequence["target_date"]
        > dates[-1]
    )


# ------------------------------------------------------------
# Target distribution
# ------------------------------------------------------------

sequence_targets = [
    int(sequence["target"])
    for sequence in temporal_sequences
]

print("\nSequence target distribution:")

print(
    pd.Series(sequence_targets)
    .value_counts()
    .sort_index()
)

print("\nALL TEMPORAL SEQUENCE CHECKS PASSED")
```

```python
# ============================================================
# PHASE 13 — TRAINING-ONLY NODE FEATURE NORMALIZATION
# ============================================================

print("=" * 70)
print("PHASE 13 FEATURE NORMALIZATION")
print("=" * 70)


# ------------------------------------------------------------
# 1. Extract all training node features
# ------------------------------------------------------------

train_feature_matrices = []

for sequence in train_sequences:

    for graph in sequence["graphs"]:

        train_feature_matrices.append(
            graph.x.numpy()
        )


train_features = np.concatenate(
    train_feature_matrices,
    axis=0
)


print("\nTraining feature matrix:")
print(train_features.shape)


# ------------------------------------------------------------
# 2. Calculate statistics ONLY from training data
# ------------------------------------------------------------

feature_mean = train_features.mean(
    axis=0
)

feature_std = train_features.std(
    axis=0
)


# Prevent division by zero
feature_std = np.where(
    feature_std < 1e-8,
    1.0,
    feature_std
)


feature_names = [
    "degree",
    "strength",
    "betweenness",
    "closeness",
    "eigenvector",
    "systemic_risk_score"
]


print("\nTraining normalization statistics:")

for name, mean, std in zip(
    feature_names,
    feature_mean,
    feature_std
):

    print(
        f"{name:20s} "
        f"mean={mean:.6f} "
        f"std={std:.6f}"
    )


# ------------------------------------------------------------
# 3. Normalization function
# ------------------------------------------------------------

def normalize_graph(
    graph,
    mean,
    std
):

    normalized_graph = graph.clone()

    normalized_x = (
        graph.x.numpy() - mean
    ) / std

    normalized_graph.x = torch.tensor(
        normalized_x,
        dtype=torch.float32
    )

    return normalized_graph


# ------------------------------------------------------------
# 4. Normalize temporal sequences
# ------------------------------------------------------------

def normalize_sequences(
    sequences,
    mean,
    std
):

    normalized_sequences = []

    for sequence in sequences:

        normalized_graphs = [
            normalize_graph(
                graph,
                mean,
                std
            )
            for graph in sequence["graphs"]
        ]

        normalized_sequences.append({
            "graphs": normalized_graphs,
            "target": sequence["target"],
            "target_date": sequence["target_date"]
        })

    return normalized_sequences


train_sequences_norm = normalize_sequences(
    train_sequences,
    feature_mean,
    feature_std
)

val_sequences_norm = normalize_sequences(
    val_sequences,
    feature_mean,
    feature_std
)

test_sequences_norm = normalize_sequences(
    test_sequences,
    feature_mean,
    feature_std
)


# ------------------------------------------------------------
# 5. Check normalization
# ------------------------------------------------------------

normalized_train_features = np.concatenate(
    [
        graph.x.numpy()
        for sequence in train_sequences_norm
        for graph in sequence["graphs"]
    ],
    axis=0
)


print("\nNormalized training statistics:")

for i, name in enumerate(feature_names):

    print(
        f"{name:20s} "
        f"mean={normalized_train_features[:, i].mean():.6f} "
        f"std={normalized_train_features[:, i].std():.6f}"
    )


# ------------------------------------------------------------
# 6. Check for invalid values
# ------------------------------------------------------------

assert not np.isnan(
    normalized_train_features
).any()

assert not np.isinf(
    normalized_train_features
).any()


# ------------------------------------------------------------
# 7. Verify validation/test use the SAME statistics
# ------------------------------------------------------------

assert len(train_sequences_norm) == len(
    train_sequences
)

assert len(val_sequences_norm) == len(
    val_sequences
)

assert len(test_sequences_norm) == len(
    test_sequences
)


print("\n" + "=" * 70)
print("NORMALIZATION COMPLETE")
print("=" * 70)

print("\nNormalization statistics were calculated from TRAINING data only.")

print(
    "Train sequences:",
    len(train_sequences_norm)
)

print(
    "Validation sequences:",
    len(val_sequences_norm)
)

print(
    "Test sequences:",
    len(test_sequences_norm)
)

print("\nALL NORMALIZATION CHECKS PASSED")
```

```python
# ============================================================
# PHASE 13 — TEMPORAL GNN MODEL
# GCN + GRU
# ============================================================

import torch
import torch.nn as nn

from torch_geometric.nn import GCNConv, global_mean_pool


# ------------------------------------------------------------
# Device
# ------------------------------------------------------------

device = torch.device(
    "cuda" if torch.cuda.is_available() else "cpu"
)

print("Device:", device)


# ------------------------------------------------------------
# Temporal GNN
# ------------------------------------------------------------

class TemporalGNN(nn.Module):

    def __init__(
        self,
        input_dim=6,
        gcn_hidden=32,
        embedding_dim=32,
        gru_hidden=32,
        dropout=0.2
    ):

        super().__init__()

        # ----------------------------------------
        # Spatial graph encoder
        # ----------------------------------------

        self.gcn1 = GCNConv(
            input_dim,
            gcn_hidden
        )

        self.gcn2 = GCNConv(
            gcn_hidden,
            embedding_dim
        )

        # ----------------------------------------
        # Temporal encoder
        # ----------------------------------------

        self.gru = nn.GRU(
            input_size=embedding_dim,
            hidden_size=gru_hidden,
            num_layers=1,
            batch_first=True
        )

        # ----------------------------------------
        # Classifier
        # ----------------------------------------

        self.classifier = nn.Sequential(
            nn.Linear(
                gru_hidden,
                16
            ),

            nn.ReLU(),

            nn.Dropout(dropout),

            nn.Linear(
                16,
                1
            )
        )


    def encode_graph(self, data):

        x = data.x.to(device)

        edge_index = data.edge_index.to(device)

        edge_weight = data.edge_weight.to(device)

        # ----------------------------------------
        # GCN layer 1
        # ----------------------------------------

        x = self.gcn1(
            x,
            edge_index,
            edge_weight
        )

        x = torch.relu(x)

        # ----------------------------------------
        # GCN layer 2
        # ----------------------------------------

        x = self.gcn2(
            x,
            edge_index,
            edge_weight
        )

        x = torch.relu(x)

        # ----------------------------------------
        # Global graph pooling
        # ----------------------------------------

        batch = torch.zeros(
            x.size(0),
            dtype=torch.long,
            device=device
        )

        graph_embedding = global_mean_pool(
            x,
            batch
        )

        return graph_embedding


    def forward(self, graphs):

        graph_embeddings = []

        # Encode each graph snapshot
        for graph in graphs:

            embedding = self.encode_graph(
                graph
            )

            graph_embeddings.append(
                embedding
            )

        # [sequence_length, 1, embedding_dim]
        temporal_input = torch.cat(
            graph_embeddings,
            dim=0
        )

        # [1, sequence_length, embedding_dim]
        temporal_input = temporal_input.unsqueeze(0)

        # GRU
        gru_output, _ = self.gru(
            temporal_input
        )

        # Last temporal state
        final_embedding = gru_output[:, -1, :]

        # Binary classification logit
        logit = self.classifier(
            final_embedding
        )

        return logit.squeeze(1)


# ------------------------------------------------------------
# Create model
# ------------------------------------------------------------

model = TemporalGNN(
    input_dim=6,
    gcn_hidden=32,
    embedding_dim=32,
    gru_hidden=32,
    dropout=0.2
).to(device)


# ------------------------------------------------------------
# Parameter summary
# ------------------------------------------------------------

total_parameters = sum(
    parameter.numel()
    for parameter in model.parameters()
)

trainable_parameters = sum(
    parameter.numel()
    for parameter in model.parameters()
    if parameter.requires_grad
)


print("\nModel:")
print(model)

print(
    "\nTotal parameters:",
    total_parameters
)

print(
    "Trainable parameters:",
    trainable_parameters
)
```

```python
# ============================================================
# PHASE 13 — TEMPORAL GNN TRAINING
# ============================================================

import copy
import random

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    average_precision_score
)


# ------------------------------------------------------------
# 1. Reproducibility
# ------------------------------------------------------------

SEED = 42

random.seed(SEED)
np.random.seed(SEED)
torch.manual_seed(SEED)


# ------------------------------------------------------------
# 2. Training configuration
# ------------------------------------------------------------

LEARNING_RATE = 0.001
WEIGHT_DECAY = 1e-4
MAX_EPOCHS = 300
PATIENCE = 40


# ------------------------------------------------------------
# 3. Fresh model
# ------------------------------------------------------------

model = TemporalGNN(
    input_dim=6,
    gcn_hidden=32,
    embedding_dim=32,
    gru_hidden=32,
    dropout=0.2
).to(device)


# ------------------------------------------------------------
# 4. Loss and optimizer
# ------------------------------------------------------------

criterion = nn.BCEWithLogitsLoss()

optimizer = torch.optim.Adam(
    model.parameters(),
    lr=LEARNING_RATE,
    weight_decay=WEIGHT_DECAY
)


# ------------------------------------------------------------
# 5. Training function
# ------------------------------------------------------------

def train_one_epoch(
    model,
    sequences
):

    model.train()

    total_loss = 0.0

    for sequence in sequences:

        optimizer.zero_grad()

        logit = model(
            sequence["graphs"]
        )

        target = torch.tensor(
            [sequence["target"]],
            dtype=torch.float32,
            device=device
        )

        loss = criterion(
            logit,
            target
        )

        loss.backward()

        torch.nn.utils.clip_grad_norm_(
            model.parameters(),
            max_norm=1.0
        )

        optimizer.step()

        total_loss += loss.item()

    return total_loss / len(sequences)


# ------------------------------------------------------------
# 6. Evaluation loss
# ------------------------------------------------------------

def evaluate_loss(
    model,
    sequences
):

    model.eval()

    total_loss = 0.0

    with torch.no_grad():

        for sequence in sequences:

            logit = model(
                sequence["graphs"]
            )

            target = torch.tensor(
                [sequence["target"]],
                dtype=torch.float32,
                device=device
            )

            loss = criterion(
                logit,
                target
            )

            total_loss += loss.item()

    return total_loss / len(sequences)


# ------------------------------------------------------------
# 7. Training loop
# ------------------------------------------------------------

history = []

best_val_loss = float("inf")

best_state = None

patience_counter = 0


print("=" * 70)
print("TRAINING TEMPORAL GNN")
print("=" * 70)

print("\nTraining samples:", len(train_sequences_norm))
print("Validation samples:", len(val_sequences_norm))
print("Maximum epochs:", MAX_EPOCHS)


for epoch in range(1, MAX_EPOCHS + 1):

    train_loss = train_one_epoch(
        model,
        train_sequences_norm
    )

    val_loss = evaluate_loss(
        model,
        val_sequences_norm
    )

    history.append({
        "epoch": epoch,
        "train_loss": train_loss,
        "val_loss": val_loss
    })


    # ----------------------------------------
    # Best model
    # ----------------------------------------

    if val_loss < best_val_loss:

        best_val_loss = val_loss

        best_state = copy.deepcopy(
            model.state_dict()
        )

        patience_counter = 0

    else:

        patience_counter += 1


    # ----------------------------------------
    # Progress
    # ----------------------------------------

    if (
        epoch == 1
        or epoch % 10 == 0
    ):

        print(
            f"Epoch {epoch:03d} | "
            f"Train Loss: {train_loss:.4f} | "
            f"Val Loss: {val_loss:.4f}"
        )


    # ----------------------------------------
    # Early stopping
    # ----------------------------------------

    if patience_counter >= PATIENCE:

        print(
            f"\nEarly stopping at epoch {epoch}"
        )

        break


# ------------------------------------------------------------
# 8. Restore best model
# ------------------------------------------------------------

if best_state is not None:

    model.load_state_dict(
        best_state
    )


print("\n" + "=" * 70)
print("TRAINING COMPLETE")
print("=" * 70)

print(
    "Best validation loss:",
    round(best_val_loss, 6)
)

print(
    "Best epoch:",
    min(
        history,
        key=lambda x: x["val_loss"]
    )["epoch"]
)

print(
    "Epochs trained:",
    len(history)
)
```

```python
# ================================================================
# CELL 3: Prepare return matrix
# ================================================================

companies = [
    "AAPL",
    "AMZN",
    "GOOGL",
    "JPM",
    "MA",
    "META",
    "MSFT",
    "NVDA",
    "TSLA",
    "V"
]

returns_phase13 = returns_df.copy()

returns_phase13["date"] = pd.to_datetime(
    returns_phase13["date"]
)

returns_phase13 = (
    returns_phase13
    .set_index("date")
    .sort_index()
)

returns_phase13 = returns_phase13[
    companies
]

print("Returns shape:", returns_phase13.shape)
print("Companies:", list(returns_phase13.columns))
print("First date:", returns_phase13.index.min())
print("Last date:", returns_phase13.index.max())
```

```python
# ================================================================
# CELL 4: Dynamic financial graph reconstruction
# ================================================================

WINDOW_SIZE = 60
STEP_SIZE = 20
CORRELATION_THRESHOLD = 0.5

graph_snapshots_13 = []

for start in range(
    0,
    len(returns_phase13) - WINDOW_SIZE + 1,
    STEP_SIZE
):

    end = start + WINDOW_SIZE

    window = returns_phase13.iloc[
        start:end
    ]

    corr_matrix = window.corr()

    graph = nx.Graph()

    # Add all companies as nodes
    for company in companies:
        graph.add_node(company)

    # Add correlation edges
    for i in range(len(companies)):

        for j in range(i + 1, len(companies)):

            company_a = companies[i]
            company_b = companies[j]

            correlation = corr_matrix.iloc[
                i,
                j
            ]

            if (
                pd.notna(correlation)
                and abs(correlation)
                >= CORRELATION_THRESHOLD
            ):

                graph.add_edge(
                    company_a,
                    company_b,
                    weight=abs(correlation),
                    distance=1 - abs(correlation),
                    correlation=correlation
                )

    snapshot_date = window.index[-1]

    graph_snapshots_13.append({
        "date": snapshot_date,
        "graph": graph,
        "correlation_matrix": corr_matrix
    })

print("=" * 70)
print("DYNAMIC GRAPH CONSTRUCTION")
print("=" * 70)

print(
    "Snapshots:",
    len(graph_snapshots_13)
)

print(
    "First date:",
    graph_snapshots_13[0]["date"]
)

print(
    "Last date:",
    graph_snapshots_13[-1]["date"]
)

print(
    "Nodes:",
    graph_snapshots_13[0]["graph"].number_of_nodes()
)

print(
    "Edges:",
    graph_snapshots_13[0]["graph"].number_of_edges()
)
```

```python
# ================================================================
# CELL 12: Training-only feature normalization
# ================================================================

all_train_features = np.concatenate(
    [
        graph.x.numpy()
        for sequence in train_sequences
        for graph in sequence["graphs"]
    ],
    axis=0
)

feature_mean = all_train_features.mean(
    axis=0
)

feature_std = all_train_features.std(
    axis=0
)

feature_std[
    feature_std == 0
] = 1.0


def normalize_sequences(
    sequences,
    mean,
    std
):

    normalized = []

    for sequence in sequences:

        new_graphs = []

        for graph in sequence["graphs"]:

            graph_copy = copy.deepcopy(
                graph
            )

            x = graph_copy.x.numpy()

            x = (
                x - mean
            ) / std

            graph_copy.x = torch.tensor(
                x,
                dtype=torch.float32
            )

            new_graphs.append(
                graph_copy
            )

        normalized.append({
            "graphs": new_graphs,
            "target": sequence["target"],
            "target_date": sequence["target_date"]
        })

    return normalized


train_sequences_norm = normalize_sequences(
    train_sequences,
    feature_mean,
    feature_std
)

val_sequences_norm = normalize_sequences(
    val_sequences,
    feature_mean,
    feature_std
)

test_sequences_norm = normalize_sequences(
    test_sequences,
    feature_mean,
    feature_std
)

print("=" * 70)
print("NORMALIZATION")
print("=" * 70)

print(
    "Training rows:",
    all_train_features.shape[0]
)

print(
    "Feature means:",
    np.round(feature_mean, 4)
)

print(
    "Feature std:",
    np.round(feature_std, 4)
)

print("\nNormalization uses TRAINING DATA ONLY")

print("\nNO DATA LEAKAGE CHECK PASSED")
```

```python
# ================================================================
# CELL 13: Small Temporal GNN baseline
# ================================================================

class TemporalGNN_Small(nn.Module):

    def __init__(
        self,
        input_dim=6,
        gcn_hidden=16,
        embedding_dim=16,
        gru_hidden=16
    ):

        super().__init__()

        self.gcn1 = GCNConv(
            input_dim,
            gcn_hidden
        )

        self.gcn2 = GCNConv(
            gcn_hidden,
            embedding_dim
        )

        self.gru = nn.GRU(
            input_size=embedding_dim,
            hidden_size=gru_hidden,
            batch_first=True
        )

        self.classifier = nn.Linear(
            gru_hidden,
            1
        )

    def encode_graph(
        self,
        graph
    ):

        x = graph.x.to(device)

        edge_index = (
            graph.edge_index.to(device)
        )

        edge_weight = (
            graph.edge_weight.to(device)
        )

        x = self.gcn1(
            x,
            edge_index,
            edge_weight
        )

        x = torch.relu(x)

        x = self.gcn2(
            x,
            edge_index,
            edge_weight
        )

        x = torch.relu(x)

        batch = torch.zeros(
            x.size(0),
            dtype=torch.long,
            device=device
        )

        graph_embedding = global_mean_pool(
            x,
            batch
        )

        return graph_embedding

    def forward(
        self,
        graphs
    ):

        embeddings = []

        for graph in graphs:

            embedding = self.encode_graph(
                graph
            )

            embeddings.append(
                embedding
            )

        sequence = torch.stack(
            embeddings,
            dim=1
        )

        output, _ = self.gru(
            sequence
        )

        last_state = output[
            :,
            -1,
            :
        ]

        logit = self.classifier(
            last_state
        )

        return logit.squeeze(-1)


small_model = TemporalGNN_Small(
    input_dim=6,
    gcn_hidden=16,
    embedding_dim=16,
    gru_hidden=16
).to(device)

parameter_count = sum(
    p.numel()
    for p in small_model.parameters()
    if p.requires_grad
)

print("=" * 70)
print("SMALL TEMPORAL GNN")
print("=" * 70)

print(
    "Trainable parameters:",
    parameter_count
)
```

```python
# ================================================================
# CELL 14: Train small Temporal GNN
# ================================================================

LEARNING_RATE = 0.001
WEIGHT_DECAY = 0.0

MAX_EPOCHS = 300
PATIENCE = 40

criterion = nn.BCEWithLogitsLoss()

optimizer = torch.optim.Adam(
    small_model.parameters(),
    lr=LEARNING_RATE,
    weight_decay=WEIGHT_DECAY
)


def train_epoch(
    model,
    sequences
):

    model.train()

    total_loss = 0.0

    for sequence in sequences:

        optimizer.zero_grad()

        logit = model(
            sequence["graphs"]
        )

        target = torch.tensor(
            [sequence["target"]],
            dtype=torch.float32,
            device=device
        )

        loss = criterion(
            logit,
            target
        )

        loss.backward()

        torch.nn.utils.clip_grad_norm_(
            model.parameters(),
            max_norm=1.0
        )

        optimizer.step()

        total_loss += loss.item()

    return (
        total_loss /
        len(sequences)
    )


def validation_loss(
    model,
    sequences
):

    model.eval()

    total_loss = 0.0

    with torch.no_grad():

        for sequence in sequences:

            logit = model(
                sequence["graphs"]
            )

            target = torch.tensor(
                [sequence["target"]],
                dtype=torch.float32,
                device=device
            )

            loss = criterion(
                logit,
                target
            )

            total_loss += loss.item()

    return (
        total_loss /
        len(sequences)
    )


history_small = []

best_val_loss = float("inf")

best_state = None

best_epoch = 0

patience_counter = 0


for epoch in range(
    1,
    MAX_EPOCHS + 1
):

    train_loss = train_epoch(
        small_model,
        train_sequences_norm
    )

    val_loss = validation_loss(
        small_model,
        val_sequences_norm
    )

    history_small.append({
        "epoch": epoch,
        "train_loss": train_loss,
        "val_loss": val_loss
    })

    if val_loss < best_val_loss:

        best_val_loss = val_loss

        best_state = copy.deepcopy(
            small_model.state_dict()
        )

        best_epoch = epoch

        patience_counter = 0

    else:

        patience_counter += 1

    if (
        epoch == 1
        or epoch % 10 == 0
    ):

        print(
            f"Epoch {epoch:03d} | "
            f"Train Loss: {train_loss:.4f} | "
            f"Val Loss: {val_loss:.4f}"
        )

    if patience_counter >= PATIENCE:

        print(
            f"\nEarly stopping at epoch {epoch}"
        )

        break


small_model.load_state_dict(
    best_state
)

print("\n" + "=" * 70)
print("TRAINING COMPLETE")
print("=" * 70)

print(
    "Best validation loss:",
    round(best_val_loss, 6)
)

print(
    "Best epoch:",
    best_epoch
)

print(
    "Epochs trained:",
    len(history_small)
)
```

```python
# ================================================================
# CELL 16: Prediction function
# ================================================================

def predict_sequences(
    model,
    sequences
):

    model.eval()

    results = []

    with torch.no_grad():

        for sequence in sequences:

            logit = model(
                sequence["graphs"]
            )

            probability = torch.sigmoid(
                logit
            ).item()

            prediction = int(
                probability >= 0.5
            )

            results.append({
                "target_date":
                    sequence["target_date"],
                "target":
                    sequence["target"],
                "probability":
                    probability,
                "prediction":
                    prediction
            })

    return pd.DataFrame(
        results
    )


train_pred_df = predict_sequences(
    small_model,
    train_sequences_norm
)

val_pred_df = predict_sequences(
    small_model,
    val_sequences_norm
)

test_pred_df = predict_sequences(
    small_model,
    test_sequences_norm
)

print("=" * 70)
print("PREDICTIONS")
print("=" * 70)

print("\nTRAIN")
display(train_pred_df)

print("\nVALIDATION")
display(val_pred_df)

print("\nTEST")
display(test_pred_df)
```

```python
# ================================================================
# PHASE 13.11
# CELL 21: Expanded Daily Dynamic Financial Graphs
# ================================================================

EXPANDED_WINDOW_SIZE = 60
EXPANDED_STEP_SIZE = 1
EXPANDED_THRESHOLD = 0.5

expanded_graph_snapshots = []

for start in range(
    0,
    len(returns_phase13) - EXPANDED_WINDOW_SIZE + 1,
    EXPANDED_STEP_SIZE
):

    end = start + EXPANDED_WINDOW_SIZE

    window = returns_phase13.iloc[
        start:end
    ]

    corr_matrix = window.corr()

    graph = nx.Graph()

    # ------------------------------------------------
    # Add all companies
    # ------------------------------------------------

    for company in companies:
        graph.add_node(company)

    # ------------------------------------------------
    # Correlation edges
    # ------------------------------------------------

    for i in range(len(companies)):

        for j in range(i + 1, len(companies)):

            company_a = companies[i]
            company_b = companies[j]

            correlation = corr_matrix.iloc[
                i,
                j
            ]

            if (
                pd.notna(correlation)
                and abs(correlation)
                >= EXPANDED_THRESHOLD
            ):

                graph.add_edge(
                    company_a,
                    company_b,
                    weight=float(
                        abs(correlation)
                    ),
                    distance=float(
                        1 - abs(correlation)
                    ),
                    correlation=float(
                        correlation
                    )
                )

    snapshot_date = window.index[-1]

    expanded_graph_snapshots.append({

        "date": snapshot_date,

        "graph": graph,

        "correlation_matrix": corr_matrix

    })


print("=" * 70)
print("EXPANDED DAILY GRAPH DATASET")
print("=" * 70)

print(
    "Snapshots:",
    len(expanded_graph_snapshots)
)

print(
    "First date:",
    expanded_graph_snapshots[0]["date"]
)

print(
    "Last date:",
    expanded_graph_snapshots[-1]["date"]
)

print(
    "Nodes per graph:",
    sorted(set(
        item["graph"].number_of_nodes()
        for item in expanded_graph_snapshots
    ))
)

edge_counts_expanded = [
    item["graph"].number_of_edges()
    for item in expanded_graph_snapshots
]

print(
    "Minimum edges:",
    min(edge_counts_expanded)
)

print(
    "Maximum edges:",
    max(edge_counts_expanded)
)

print(
    "Average edges:",
    round(
        np.mean(edge_counts_expanded),
        2
    )
)
```

```python
# ================================================================
# CELL 23: Expanded graph-level feature extraction
# ================================================================

expanded_graph_features = []

for snapshot in expanded_graph_snapshots:

    date = snapshot["date"]
    graph = snapshot["graph"]

    # ------------------------------------------------
    # Basic topology
    # ------------------------------------------------

    nodes = graph.number_of_nodes()
    edges = graph.number_of_edges()

    density = nx.density(graph)

    average_degree = (
        np.mean([
            degree
            for _, degree
            in graph.degree()
        ])
    )

    # ------------------------------------------------
    # Weighted strength
    # ------------------------------------------------

    strengths = dict(
        graph.degree(
            weight="weight"
        )
    )

    average_strength = np.mean(
        list(strengths.values())
    )

    # ------------------------------------------------
    # Clustering
    # ------------------------------------------------

    average_clustering = nx.average_clustering(
        graph,
        weight="weight"
    )

    # ------------------------------------------------
    # Connected components
    # ------------------------------------------------

    connected_components = (
        nx.number_connected_components(
            graph
        )
    )

    # ------------------------------------------------
    # Centralization
    # ------------------------------------------------

    degree_values = np.array(
        list(
            dict(
                graph.degree()
            ).values()
        ),
        dtype=float
    )

    degree_concentration = (
        np.std(degree_values)
        if len(degree_values) > 0
        else 0.0
    )

    # ------------------------------------------------
    # Record
    # ------------------------------------------------

    expanded_graph_features.append({

        "date": date,

        "nodes": nodes,

        "edges": edges,

        "density": density,

        "average_degree": average_degree,

        "average_strength": average_strength,

        "average_clustering":
            average_clustering,

        "connected_components":
            connected_components,

        "degree_concentration":
            degree_concentration
    })


expanded_graph_features_df = pd.DataFrame(
    expanded_graph_features
)

print("=" * 70)
print("EXPANDED GRAPH FEATURES")
print("=" * 70)

print(
    "Shape:",
    expanded_graph_features_df.shape
)

display(
    expanded_graph_features_df.head()
)

print("\nMissing values:")

print(
    expanded_graph_features_df.isna()
    .sum()
)
```

```python
# ================================================================
# CELL 24: Daily market features
# ================================================================

market_features = []

for date in expanded_dates:

    # --------------------------------------------
    # Return vector on this date
    # --------------------------------------------

    returns_row = returns_phase13.loc[
        date
    ]

    return_values = (
        returns_row
        .astype(float)
        .values
    )

    # --------------------------------------------
    # Market return
    # --------------------------------------------

    market_return = np.mean(
        return_values
    )

    # --------------------------------------------
    # Cross-sectional volatility
    # --------------------------------------------

    market_volatility = np.std(
        return_values
    )

    # --------------------------------------------
    # Return dispersion
    # --------------------------------------------

    return_dispersion = np.std(
        return_values
    )

    # --------------------------------------------
    # Fraction of negative returns
    # --------------------------------------------

    negative_fraction = np.mean(
        return_values < 0
    )

    market_features.append({

        "date": date,

        "market_return":
            market_return,

        "market_volatility":
            market_volatility,

        "return_dispersion":
            return_dispersion,

        "negative_fraction":
            negative_fraction
    })


market_features_df = pd.DataFrame(
    market_features
)

print("=" * 70)
print("DAILY MARKET FEATURES")
print("=" * 70)

print(
    "Shape:",
    market_features_df.shape
)

display(
    market_features_df.head()
)
```

```python
# ================================================================
# CELL 25: Market drawdown
# ================================================================

market_features_df = (
    market_features_df
    .sort_values("date")
    .reset_index(drop=True)
)

market_features_df["market_index"] = (
    1
    + market_features_df[
        "market_return"
    ]
).cumprod()

market_features_df[
    "running_peak"
] = market_features_df[
    "market_index"
].cummax()

market_features_df[
    "drawdown"
] = (
    market_features_df[
        "market_index"
    ]
    /
    market_features_df[
        "running_peak"
    ]
    - 1
)

print(
    market_features_df[
        [
            "date",
            "market_return",
            "market_volatility",
            "drawdown"
        ]
    ].head(10)
)
```

```python
# ================================================================
# CELL 26: Expanded financial graph state
# ================================================================

expanded_state_df = (
    expanded_graph_features_df
    .merge(
        market_features_df,
        on="date",
        how="inner"
    )
    .sort_values("date")
    .reset_index(drop=True)
)

print("=" * 70)
print("EXPANDED FINANCIAL GRAPH STATE")
print("=" * 70)

print(
    "Shape:",
    expanded_state_df.shape
)

print(
    "Date range:",
    expanded_state_df["date"].min(),
    "→",
    expanded_state_df["date"].max()
)

print("\nColumns:")

print(
    expanded_state_df.columns.tolist()
)

print("\nMissing values:")

print(
    expanded_state_df.isna().sum()
)
```

```python
# ================================================================
# PHASE 13.12
# CELL 27 — Correct Future Stress Variables
# ================================================================

FORECAST_HORIZON = 20

future_df = (
    expanded_state_df
    .copy()
    .sort_values("date")
    .reset_index(drop=True)
)

n_obs = len(future_df)

# ------------------------------------------------
# Initialize future variables with NaN
# ------------------------------------------------

future_df["future_return_20d"] = np.nan
future_df["future_max_drawdown_20d"] = np.nan
future_df["future_volatility_20d"] = np.nan
future_df["future_negative_fraction_20d"] = np.nan

market_index = (
    future_df["market_index"]
    .astype(float)
    .values
)

market_returns = (
    future_df["market_return"]
    .astype(float)
    .values
)

# ------------------------------------------------
# Only calculate values when a COMPLETE
# 20-trading-day future window exists.
#
# Therefore:
#
# t -> t+20
#
# is required for every usable observation.
# ------------------------------------------------

for i in range(
    n_obs - FORECAST_HORIZON
):

    future_start = i + 1
    future_end = i + FORECAST_HORIZON

    future_indices = np.arange(
        future_start,
        future_end + 1
    )

    # --------------------------------------------
    # Future cumulative return
    # --------------------------------------------

    future_df.loc[
        i,
        "future_return_20d"
    ] = (
        market_index[future_end]
        /
        market_index[i]
        - 1
    )

    # --------------------------------------------
    # Future maximum drawdown
    #
    # Measured relative to the market level
    # at prediction time t.
    # --------------------------------------------

    future_path = market_index[
        future_indices
    ]

    future_drawdowns = (
        future_path
        /
        market_index[i]
        - 1
    )

    future_df.loc[
        i,
        "future_max_drawdown_20d"
    ] = future_drawdowns.min()

    # --------------------------------------------
    # Future realized volatility
    # --------------------------------------------

    future_path_returns = market_returns[
        future_indices
    ]

    future_df.loc[
        i,
        "future_volatility_20d"
    ] = np.std(
        future_path_returns
    )

    # --------------------------------------------
    # Future negative-return fraction
    # --------------------------------------------

    future_df.loc[
        i,
        "future_negative_fraction_20d"
    ] = np.mean(
        future_path_returns < 0
    )


print("=" * 70)
print("CORRECTED FUTURE STRESS VARIABLES")
print("=" * 70)

print(
    "Forecast horizon:",
    FORECAST_HORIZON,
    "trading days"
)

print(
    "Total observations:",
    len(future_df)
)

print("\nMissing values:")

print(
    future_df[
        [
            "future_return_20d",
            "future_max_drawdown_20d",
            "future_volatility_20d",
            "future_negative_fraction_20d"
        ]
    ].isna().sum()
)

print("\nExpected missing values:")

print(
    "Last",
    FORECAST_HORIZON,
    "observations should be NaN."
)

print("\nFuture variable summary:")

display(
    future_df[
        [
            "future_return_20d",
            "future_max_drawdown_20d",
            "future_volatility_20d",
            "future_negative_fraction_20d"
        ]
    ].describe()
)
```

```python
# ================================================================
# PHASE 13.14
# CELL 41 — DAILY ROLLING FINANCIAL GRAPHS
# ================================================================

import networkx as nx
import numpy as np
import pandas as pd

WINDOW_SIZE = 60
CORRELATION_THRESHOLD = 0.5

companies = [
    "AAPL",
    "AMZN",
    "GOOGL",
    "JPM",
    "MA",
    "META",
    "MSFT",
    "NVDA",
    "TSLA",
    "V"
]

returns_daily = returns_df.copy()

returns_daily["date"] = pd.to_datetime(
    returns_daily["date"]
)

returns_daily = (
    returns_daily
    .sort_values("date")
    .reset_index(drop=True)
)

returns_daily = returns_daily[
    ["date"] + companies
].copy()

print("=" * 70)
print("BUILDING DAILY ROLLING FINANCIAL GRAPHS")
print("=" * 70)

daily_graph_snapshots = []

for end_idx in range(
    WINDOW_SIZE - 1,
    len(returns_daily)
):

    start_idx = (
        end_idx
        - WINDOW_SIZE
        + 1
    )

    window = returns_daily.iloc[
        start_idx:end_idx + 1
    ]

    corr_matrix = window[
        companies
    ].corr()

    graph = nx.Graph()

    # ------------------------------------------------
    # Add all companies
    # ------------------------------------------------

    graph.add_nodes_from(companies)

    # ------------------------------------------------
    # Add correlation edges
    # ------------------------------------------------

    for i in range(len(companies)):

        for j in range(i + 1, len(companies)):

            company_a = companies[i]
            company_b = companies[j]

            correlation = corr_matrix.loc[
                company_a,
                company_b
            ]

            if (
                pd.notna(correlation)
                and
                abs(correlation)
                >= CORRELATION_THRESHOLD
            ):

                graph.add_edge(
                    company_a,
                    company_b,
                    weight=float(
                        abs(correlation)
                    ),
                    correlation=float(
                        correlation
                    ),
                    distance=float(
                        1 - abs(correlation)
                    )
                )

    snapshot_date = returns_daily.loc[
        end_idx,
        "date"
    ]

    daily_graph_snapshots.append({

        "date": snapshot_date,

        "graph": graph,

        "correlation_matrix":
            corr_matrix

    })

print(
    "Daily graph snapshots:",
    len(daily_graph_snapshots)
)

print(
    "First graph date:",
    daily_graph_snapshots[0]["date"]
)

print(
    "Last graph date:",
    daily_graph_snapshots[-1]["date"]
)
```

```python
# ================================================================
# CELL 43 — DAILY NODE FEATURES
# ================================================================

print("=" * 70)
print("BUILDING DAILY NODE FEATURES")
print("=" * 70)

# ------------------------------------------------
# Prepare systemic risk data
# ------------------------------------------------

systemic_daily = systemic_risk_df.copy()

systemic_daily["date"] = pd.to_datetime(
    systemic_daily["date"]
)

systemic_lookup = {}

for date, group in systemic_daily.groupby("date"):

    systemic_lookup[date] = (
        group
        .set_index("company")[
            "systemic_risk_score"
        ]
        .to_dict()
    )

# ------------------------------------------------
# Returns lookup
# ------------------------------------------------

returns_lookup = (
    returns_daily
    .set_index("date")
)

# ------------------------------------------------
# Feature names
# ------------------------------------------------

node_feature_names = [
    "degree",
    "strength",
    "betweenness",
    "closeness",
    "eigenvector",
    "clustering",
    "systemic_risk",
    "return",
    "volatility",
    "negative_return"
]

daily_node_features = {}

for snapshot in daily_graph_snapshots:

    date = snapshot["date"]
    graph = snapshot["graph"]

    # ------------------------------------------------
    # Network features
    # ------------------------------------------------

    degree = dict(
        graph.degree()
    )

    strength = dict(
        graph.degree(
            weight="weight"
        )
    )

    betweenness = nx.betweenness_centrality(
        graph,
        weight="distance"
    )

    closeness = nx.closeness_centrality(
        graph,
        distance="distance"
    )

    try:

        eigenvector = nx.eigenvector_centrality_numpy(
            graph,
            weight="weight"
        )

    except Exception:

        eigenvector = {
            company: 0.0
            for company in companies
        }

    clustering = nx.clustering(
        graph,
        weight="weight"
    )

    systemic_scores = systemic_lookup.get(
        date,
        {}
    )

    return_values = (
        returns_lookup.loc[
            date,
            companies
        ]
        if date in returns_lookup.index
        else pd.Series(
            0.0,
            index=companies
        )
    )

    # ------------------------------------------------
    # Rolling local volatility
    # ------------------------------------------------

    end_position = returns_daily.index[
        returns_daily["date"] == date
    ]

    if len(end_position) == 0:

        local_volatility = pd.Series(
            0.0,
            index=companies
        )

    else:

        end_position = int(
            end_position[0]
        )

        start_position = max(
            0,
            end_position - 19
        )

        local_window = returns_daily.iloc[
            start_position:
            end_position + 1
        ][companies]

        local_volatility = (
            local_window
            .std()
        )

    feature_matrix = []

    for company in companies:

        current_return = float(
            return_values[company]
        )

        current_volatility = float(
            local_volatility[company]
        )

        negative_return = float(
            current_return < 0
        )

        feature_matrix.append([

            float(degree.get(
                company,
                0.0
            )),

            float(strength.get(
                company,
                0.0
            )),

            float(betweenness.get(
                company,
                0.0
            )),

            float(closeness.get(
                company,
                0.0
            )),

            float(eigenvector.get(
                company,
                0.0
            )),

            float(clustering.get(
                company,
                0.0
            )),

            float(systemic_scores.get(
                company,
                0.0
            )),

            current_return,

            current_volatility,

            negative_return

        ])

    daily_node_features[date] = np.asarray(
        feature_matrix,
        dtype=np.float32
    )

print(
    "Feature snapshots:",
    len(daily_node_features)
)

print(
    "Features per node:",
    len(node_feature_names)
)

print(
    "Nodes per snapshot:",
    daily_node_features[
        graph_dates[0]
    ].shape[0]
)

print(
    "Feature shape:",
    daily_node_features[
        graph_dates[0]
    ].shape
)
```

```python
# ================================================================
# PHASE 13.15
# CELL 47 — CONVERT DAILY GRAPHS TO PYTORCH GEOMETRIC
# ================================================================

import torch
import torch.nn as nn
from torch_geometric.data import Data

print("=" * 70)
print("CONVERTING TEMPORAL GRAPHS TO PYTORCH GEOMETRIC")
print("=" * 70)

device = torch.device(
    "cuda" if torch.cuda.is_available() else "cpu"
)

print("Device:", device)

date_to_features = daily_node_features

pyg_temporal_sequences = []

for sequence in temporal_sequences:

    pyg_graphs = []

    for date, graph_nx in zip(
        sequence["dates"],
        sequence["graphs"]
    ):

        # ------------------------------------------------
        # Node feature matrix
        # ------------------------------------------------

        x = torch.tensor(
            date_to_features[date],
            dtype=torch.float32
        )

        # ------------------------------------------------
        # Node ordering
        # ------------------------------------------------

        node_list = companies

        node_to_idx = {
            node: idx
            for idx, node in enumerate(node_list)
        }

        # ------------------------------------------------
        # Edges
        # ------------------------------------------------

        edge_pairs = []
        edge_weights = []

        for u, v, attrs in graph_nx.edges(
            data=True
        ):

            ui = node_to_idx[u]
            vi = node_to_idx[v]

            weight = float(
                attrs.get(
                    "weight",
                    1.0
                )
            )

            edge_pairs.append(
                [ui, vi]
            )

            edge_pairs.append(
                [vi, ui]
            )

            edge_weights.append(weight)
            edge_weights.append(weight)

        # ------------------------------------------------
        # Edge tensor
        # ------------------------------------------------

        if len(edge_pairs) > 0:

            edge_index = torch.tensor(
                edge_pairs,
                dtype=torch.long
            ).t().contiguous()

            edge_weight = torch.tensor(
                edge_weights,
                dtype=torch.float32
            )

        else:

            edge_index = torch.empty(
                (2, 0),
                dtype=torch.long
            )

            edge_weight = torch.empty(
                (0,),
                dtype=torch.float32
            )

        graph_data = Data(
            x=x,
            edge_index=edge_index,
            edge_weight=edge_weight
        )

        graph_data.date = str(date)

        pyg_graphs.append(
            graph_data
        )

    pyg_temporal_sequences.append({

        "graphs": pyg_graphs,

        "target_date":
            sequence["target_date"],

        "target":
            sequence["target"]

    })

print(
    "Temporal sequences:",
    len(pyg_temporal_sequences)
)

print(
    "Sequence length:",
    len(
        pyg_temporal_sequences[0]["graphs"]
    )
)

print(
    "Nodes per graph:",
    pyg_temporal_sequences[0][
        "graphs"
    ][0].num_nodes
)

print(
    "Features per node:",
    pyg_temporal_sequences[0][
        "graphs"
    ][0].num_node_features
)

print(
    "Edges in first graph:",
    pyg_temporal_sequences[0][
        "graphs"
    ][0].edge_index.shape[1]
)
```

```python
# ================================================================
# CELL 49 — CHRONOLOGICAL TEMPORAL SPLIT
# ================================================================

print("=" * 70)
print("CHRONOLOGICAL TEMPORAL SPLIT")
print("=" * 70)

sequence_dates = pd.to_datetime([
    sequence["target_date"]
    for sequence in pyg_temporal_sequences
])

sequence_order = np.argsort(
    sequence_dates.values
)

pyg_temporal_sequences = [
    pyg_temporal_sequences[i]
    for i in sequence_order
]

sequence_dates = pd.to_datetime([
    sequence["target_date"]
    for sequence in pyg_temporal_sequences
])

n_sequences = len(
    pyg_temporal_sequences
)

train_end = int(
    n_sequences * 0.60
)

validation_end = int(
    n_sequences * 0.80
)

train_sequences = pyg_temporal_sequences[
    :train_end
]

validation_sequences = pyg_temporal_sequences[
    train_end:validation_end
]

test_sequences = pyg_temporal_sequences[
    validation_end:
]

print(
    "Total sequences:",
    n_sequences
)

print(
    "Training:",
    len(train_sequences)
)

print(
    "Validation:",
    len(validation_sequences)
)

print(
    "Test:",
    len(test_sequences)
)

def target_counts(sequences):

    positives = sum(
        s["target"]
        for s in sequences
    )

    negatives = (
        len(sequences)
        - positives
    )

    return positives, negatives


train_pos, train_neg = target_counts(
    train_sequences
)

val_pos, val_neg = target_counts(
    validation_sequences
)

test_pos, test_neg = target_counts(
    test_sequences
)

print("\nTarget distribution:")

print(
    "TRAIN:",
    train_pos,
    "positive /",
    train_neg,
    "negative"
)

print(
    "VALIDATION:",
    val_pos,
    "positive /",
    val_neg,
    "negative"
)

print(
    "TEST:",
    test_pos,
    "positive /",
    test_neg,
    "negative"
)

print("\nPrediction-date ranges:")

print(
    "TRAIN:",
    train_sequences[0]["target_date"],
    "→",
    train_sequences[-1]["target_date"]
)

print(
    "VALIDATION:",
    validation_sequences[0]["target_date"],
    "→",
    validation_sequences[-1]["target_date"]
)

print(
    "TEST:",
    test_sequences[0]["target_date"],
    "→",
    test_sequences[-1]["target_date"]
)

assert (
    train_sequences[-1]["target_date"]
    <
    validation_sequences[0]["target_date"]
)

assert (
    validation_sequences[-1]["target_date"]
    <
    test_sequences[0]["target_date"]
)

print(
    "\nCHRONOLOGICAL SPLIT CHECKS PASSED"
)
```

```python
# ================================================================
# CELL 51 — APPLY TRAINING-ONLY NORMALIZATION
# ================================================================

def normalize_sequences(
    sequences,
    mean,
    std
):

    normalized_sequences = []

    for sequence in sequences:

        normalized_graphs = []

        for graph in sequence["graphs"]:

            graph_copy = graph.clone()

            x = (
                graph_copy.x
                .detach()
                .cpu()
                .numpy()
            )

            x_normalized = (
                x - mean
            ) / std

            graph_copy.x = torch.tensor(
                x_normalized,
                dtype=torch.float32
            )

            normalized_graphs.append(
                graph_copy
            )

        normalized_sequences.append({

            "graphs": normalized_graphs,

            "target_date":
                sequence["target_date"],

            "target":
                sequence["target"]

        })

    return normalized_sequences


# ------------------------------------------------
# Apply the SAME training statistics to all splits
# ------------------------------------------------

train_sequences_norm = normalize_sequences(
    train_sequences,
    feature_mean,
    feature_std
)

validation_sequences_norm = normalize_sequences(
    validation_sequences,
    feature_mean,
    feature_std
)

test_sequences_norm = normalize_sequences(
    test_sequences,
    feature_mean,
    feature_std
)


print("=" * 70)
print("NORMALIZATION COMPLETE")
print("=" * 70)

print(
    "Training sequences:",
    len(train_sequences_norm)
)

print(
    "Validation sequences:",
    len(validation_sequences_norm)
)

print(
    "Test sequences:",
    len(test_sequences_norm)
)
```

```python
# CELL 56 — GRAPH ENCODER

class GraphEncoder(nn.Module):

    def __init__(
        self,
        input_dim=10,
        hidden_dim=32,
        output_dim=32,
        dropout=0.2
    ):
        super().__init__()

        self.conv1 = GCNConv(
            input_dim,
            hidden_dim
        )

        self.conv2 = GCNConv(
            hidden_dim,
            output_dim
        )

        self.dropout = nn.Dropout(dropout)

    def forward(self, x, edge_index, edge_weight=None):

        # First graph convolution
        x = self.conv1(
            x,
            edge_index,
            edge_weight=edge_weight
        )

        x = F.relu(x)

        x = self.dropout(x)

        # Second graph convolution
        x = self.conv2(
            x,
            edge_index,
            edge_weight=edge_weight
        )

        x = F.relu(x)

        return x
```

```python
# CELL 57 — GRAPH-LEVEL REPRESENTATION

class GraphSystemicEncoder(nn.Module):

    def __init__(
        self,
        input_dim=10,
        hidden_dim=32,
        output_dim=32,
        dropout=0.2
    ):
        super().__init__()

        self.graph_encoder = GraphEncoder(
            input_dim=input_dim,
            hidden_dim=hidden_dim,
            output_dim=output_dim,
            dropout=dropout
        )

    def forward(self, graph):

        node_embeddings = self.graph_encoder(
            graph.x,
            graph.edge_index,
            graph.edge_weight
        )

        # All nodes belong to one graph
        batch = torch.zeros(
            graph.num_nodes,
            dtype=torch.long,
            device=graph.x.device
        )

        graph_embedding = global_mean_pool(
            node_embeddings,
            batch
        )

        return graph_embedding.squeeze(0)
```

```python
# CELL 58 — TEMPORAL ATTENTION

class TemporalAttention(nn.Module):

    def __init__(
        self,
        input_dim=32,
        attention_dim=32
    ):
        super().__init__()

        self.projection = nn.Linear(
            input_dim,
            attention_dim
        )

        self.score = nn.Linear(
            attention_dim,
            1
        )

    def forward(self, sequence):

        # sequence shape:
        # [time, embedding_dim]

        hidden = torch.tanh(
            self.projection(sequence)
        )

        scores = self.score(hidden)

        # [time]
        scores = scores.squeeze(-1)

        attention_weights = torch.softmax(
            scores,
            dim=0
        )

        # Weighted temporal representation
        context = torch.sum(
            sequence * attention_weights.unsqueeze(-1),
            dim=0
        )

        return context, attention_weights
```

```python
# CELL 59 — FULL TEMPORAL GNN

class FinGraphTemporalGNN(nn.Module):

    def __init__(
        self,
        node_feature_dim=10,
        gnn_hidden_dim=32,
        graph_embedding_dim=32,
        attention_dim=32,
        dropout=0.2
    ):
        super().__init__()

        self.graph_encoder = GraphEncoder(
            input_dim=node_feature_dim,
            hidden_dim=gnn_hidden_dim,
            output_dim=graph_embedding_dim,
            dropout=dropout
        )

        self.temporal_attention = TemporalAttention(
            input_dim=graph_embedding_dim,
            attention_dim=attention_dim
        )

        self.risk_head = nn.Sequential(

            nn.Linear(
                graph_embedding_dim,
                16
            ),

            nn.ReLU(),

            nn.Dropout(dropout),

            nn.Linear(
                16,
                1
            )
        )

    def encode_graph(self, graph):

        node_embeddings = self.graph_encoder(
            graph.x,
            graph.edge_index,
            graph.edge_weight
        )

        batch = torch.zeros(
            graph.num_nodes,
            dtype=torch.long,
            device=graph.x.device
        )

        graph_embedding = global_mean_pool(
            node_embeddings,
            batch
        )

        return graph_embedding.squeeze(0)

    def forward(self, graphs):

        graph_embeddings = []

        for graph in graphs:

            graph_embedding = self.encode_graph(
                graph
            )

            graph_embeddings.append(
                graph_embedding
            )

        # [time, embedding_dim]
        temporal_sequence = torch.stack(
            graph_embeddings,
            dim=0
        )

        temporal_context, attention_weights = (
            self.temporal_attention(
                temporal_sequence
            )
        )

        logits = self.risk_head(
            temporal_context
        )

        return logits.squeeze(), attention_weights
```

```python
# CELL 68 — TRAINING FUNCTION

def train_one_epoch(
    model,
    sequences,
    optimizer,
    criterion,
    device
):
    
    model.train()
    
    total_loss = 0.0
    correct = 0
    total = 0
    
    for sequence in sequences:
        
        graphs = [
            graph.to(device)
            for graph in sequence["graphs"]
        ]
        
        target = torch.tensor(
            float(sequence["target"]),
            dtype=torch.float32,
            device=device
        )
        
        optimizer.zero_grad()
        
        logits, _ = model(graphs)
        
        loss = criterion(
            logits.reshape(1),
            target.reshape(1)
        )
        
        loss.backward()
        
        # Prevent unstable gradient updates
        torch.nn.utils.clip_grad_norm_(
            model.parameters(),
            max_norm=1.0
        )
        
        optimizer.step()
        
        total_loss += loss.item()
        
        probability = torch.sigmoid(logits).item()
        prediction = int(probability >= 0.5)
        
        correct += int(
            prediction == int(target.item())
        )
        
        total += 1
    
    average_loss = total_loss / total
    accuracy = correct / total
    
    return average_loss, accuracy
```

```python
# CELL 69 — VALIDATION FUNCTION

@torch.no_grad()
def evaluate_loss(
    model,
    sequences,
    criterion,
    device
):
    
    model.eval()
    
    total_loss = 0.0
    correct = 0
    total = 0
    
    for sequence in sequences:
        
        graphs = [
            graph.to(device)
            for graph in sequence["graphs"]
        ]
        
        target = torch.tensor(
            float(sequence["target"]),
            dtype=torch.float32,
            device=device
        )
        
        logits, _ = model(graphs)
        
        loss = criterion(
            logits.reshape(1),
            target.reshape(1)
        )
        
        total_loss += loss.item()
        
        probability = torch.sigmoid(logits).item()
        prediction = int(probability >= 0.5)
        
        correct += int(
            prediction == int(target.item())
        )
        
        total += 1
    
    average_loss = total_loss / total
    accuracy = correct / total
    
    return average_loss, accuracy
```

```python
# CELL 70 — PREDICTION FUNCTION

@torch.no_grad()
def predict_sequences(
    model,
    sequences,
    device
):
    
    model.eval()
    
    probabilities = []
    targets = []
    dates = []
    attention_weights = []
    
    for sequence in sequences:
        
        graphs = [
            graph.to(device)
            for graph in sequence["graphs"]
        ]
        
        logits, attention = model(graphs)
        
        probability = torch.sigmoid(
            logits
        ).item()
        
        probabilities.append(probability)
        targets.append(int(sequence["target"]))
        dates.append(sequence["target_date"])
        
        attention_weights.append(
            attention.detach().cpu().numpy()
        )
    
    return {
        "probabilities": np.array(probabilities),
        "targets": np.array(targets),
        "dates": np.array(dates),
        "attention": np.array(attention_weights)
    }
```

```python
# CELL 72 — TRAINING CONFIGURATION

NUM_EPOCHS = 100

PATIENCE = 12

best_validation_loss = float("inf")

epochs_without_improvement = 0

training_history = []

best_model_state = None

print("=" * 70)
print("TRAINING CONFIGURATION")
print("=" * 70)

print("Maximum epochs:", NUM_EPOCHS)
print("Early stopping patience:", PATIENCE)
print("Learning rate:", 0.001)
print("Weight decay:", 1e-4)
print("Gradient clipping:", 1.0)
```

```python
# CELL 73 — TRAIN TEMPORAL GNN

print("=" * 70)
print("TRAINING FINGRAPH TEMPORAL GNN")
print("=" * 70)

for epoch in range(1, NUM_EPOCHS + 1):
    
    train_loss, train_accuracy = train_one_epoch(
        model,
        train_sequences_norm,
        optimizer,
        criterion,
        device
    )
    
    validation_loss, validation_accuracy = evaluate_loss(
        model,
        validation_sequences_norm,
        criterion,
        device
    )
    
    training_history.append({
        "epoch": epoch,
        "train_loss": train_loss,
        "validation_loss": validation_loss,
        "train_accuracy": train_accuracy,
        "validation_accuracy": validation_accuracy
    })
    
    # Save best model based ONLY on validation loss
    if validation_loss < best_validation_loss:
        
        best_validation_loss = validation_loss
        
        epochs_without_improvement = 0
        
        best_model_state = {
            key: value.detach().cpu().clone()
            for key, value in model.state_dict().items()
        }
        
        print(
            f"Epoch {epoch:03d} | "
            f"Train Loss: {train_loss:.4f} | "
            f"Val Loss: {validation_loss:.4f} | "
            f"Train Acc: {train_accuracy:.4f} | "
            f"Val Acc: {validation_accuracy:.4f} | "
            f"* BEST"
        )
    
    else:
        
        epochs_without_improvement += 1
        
        print(
            f"Epoch {epoch:03d} | "
            f"Train Loss: {train_loss:.4f} | "
            f"Val Loss: {validation_loss:.4f} | "
            f"Train Acc: {train_accuracy:.4f} | "
            f"Val Acc: {validation_accuracy:.4f}"
        )
    
    if epochs_without_improvement >= PATIENCE:
        
        print(
            f"\nEarly stopping at epoch {epoch}."
        )
        
        break

print("\n" + "=" * 70)
print("TRAINING COMPLETE")
print("=" * 70)

print(
    "Best validation loss:",
    round(best_validation_loss, 6)
)

print(
    "Best epoch:",
    min(
        training_history,
        key=lambda x: x["validation_loss"]
    )["epoch"]
)
```

```python
# CELL 94 — GCN + GRU BASELINE GRAPH ENCODER

class BaselineGraphEncoder(nn.Module):

    def __init__(
        self,
        input_dim=10,
        hidden_dim=32,
        output_dim=32,
        dropout=0.2
    ):
        super().__init__()

        self.conv1 = GCNConv(
            input_dim,
            hidden_dim
        )

        self.conv2 = GCNConv(
            hidden_dim,
            output_dim
        )

        self.dropout = nn.Dropout(dropout)

    def forward(
        self,
        x,
        edge_index,
        edge_weight=None
    ):

        x = self.conv1(
            x,
            edge_index,
            edge_weight=edge_weight
        )

        x = F.relu(x)

        x = self.dropout(x)

        x = self.conv2(
            x,
            edge_index,
            edge_weight=edge_weight
        )

        x = F.relu(x)

        return x
```

```python
# CELL 95 — GCN + GRU BASELINE

class GCNGRUBaseline(nn.Module):

    def __init__(
        self,
        node_feature_dim=10,
        gnn_hidden_dim=32,
        graph_embedding_dim=32,
        gru_hidden_dim=32,
        dropout=0.2
    ):
        super().__init__()

        self.graph_encoder = BaselineGraphEncoder(
            input_dim=node_feature_dim,
            hidden_dim=gnn_hidden_dim,
            output_dim=graph_embedding_dim,
            dropout=dropout
        )

        self.gru = nn.GRU(
            input_size=graph_embedding_dim,
            hidden_size=gru_hidden_dim,
            batch_first=True
        )

        self.dropout = nn.Dropout(dropout)

        self.classifier = nn.Sequential(

            nn.Linear(
                gru_hidden_dim,
                16
            ),

            nn.ReLU(),

            nn.Dropout(dropout),

            nn.Linear(
                16,
                1
            )
        )

    def encode_graph(self, graph):

        node_embeddings = self.graph_encoder(
            graph.x,
            graph.edge_index,
            graph.edge_weight
        )

        batch = torch.zeros(
            graph.num_nodes,
            dtype=torch.long,
            device=graph.x.device
        )

        graph_embedding = global_mean_pool(
            node_embeddings,
            batch
        )

        return graph_embedding.squeeze(0)

    def forward(self, graphs):

        graph_embeddings = []

        for graph in graphs:

            embedding = self.encode_graph(
                graph
            )

            graph_embeddings.append(
                embedding
            )

        # [1, time, embedding_dim]
        sequence = torch.stack(
            graph_embeddings,
            dim=0
        ).unsqueeze(0)

        gru_output, hidden = self.gru(
            sequence
        )

        # Final temporal state
        temporal_representation = (
            gru_output[:, -1, :]
            .squeeze(0)
        )

        temporal_representation = self.dropout(
            temporal_representation
        )

        logits = self.classifier(
            temporal_representation
        )

        return logits.squeeze()
```

```python
# CELL 99 — BASELINE TRAINING FUNCTION

def train_baseline_one_epoch(
    model,
    sequences,
    optimizer,
    criterion,
    device
):

    model.train()

    total_loss = 0.0
    correct = 0
    total = 0

    for sequence in sequences:

        graphs = [
            graph.to(device)
            for graph in sequence["graphs"]
        ]

        target = torch.tensor(
            float(sequence["target"]),
            dtype=torch.float32,
            device=device
        )

        optimizer.zero_grad()

        logits = model(graphs)

        loss = criterion(
            logits.reshape(1),
            target.reshape(1)
        )

        loss.backward()

        torch.nn.utils.clip_grad_norm_(
            model.parameters(),
            max_norm=1.0
        )

        optimizer.step()

        total_loss += loss.item()

        probability = torch.sigmoid(
            logits
        ).item()

        prediction = int(
            probability >= 0.5
        )

        correct += int(
            prediction == int(target.item())
        )

        total += 1

    return (
        total_loss / total,
        correct / total
    )
```

```python
# CELL 100 — BASELINE VALIDATION FUNCTION

@torch.no_grad()
def evaluate_baseline_loss(
    model,
    sequences,
    criterion,
    device
):

    model.eval()

    total_loss = 0.0
    correct = 0
    total = 0

    for sequence in sequences:

        graphs = [
            graph.to(device)
            for graph in sequence["graphs"]
        ]

        target = torch.tensor(
            float(sequence["target"]),
            dtype=torch.float32,
            device=device
        )

        logits = model(graphs)

        loss = criterion(
            logits.reshape(1),
            target.reshape(1)
        )

        total_loss += loss.item()

        probability = torch.sigmoid(
            logits
        ).item()

        prediction = int(
            probability >= 0.5
        )

        correct += int(
            prediction == int(target.item())
        )

        total += 1

    return (
        total_loss / total,
        correct / total
    )
```

```python
# CELL 101 — BASELINE PREDICTION FUNCTION

@torch.no_grad()
def predict_baseline(
    model,
    sequences,
    device
):

    model.eval()

    probabilities = []
    targets = []
    dates = []

    for sequence in sequences:

        graphs = [
            graph.to(device)
            for graph in sequence["graphs"]
        ]

        logits = model(graphs)

        probability = torch.sigmoid(
            logits
        ).item()

        probabilities.append(
            probability
        )

        targets.append(
            int(sequence["target"])
        )

        dates.append(
            sequence["target_date"]
        )

    return {
        "probabilities": np.array(
            probabilities
        ),
        "targets": np.array(
            targets
        ),
        "dates": np.array(
            dates
        )
    }
```

```python
# CELL 102 — TRAIN GCN + GRU BASELINE

BASELINE_EPOCHS = 100
BASELINE_PATIENCE = 12

baseline_best_loss = float("inf")
baseline_epochs_without_improvement = 0
baseline_best_state = None
baseline_history = []

print("=" * 70)
print("TRAINING GCN + GRU BASELINE")
print("=" * 70)

for epoch in range(1, BASELINE_EPOCHS + 1):

    train_loss, train_accuracy = (
        train_baseline_one_epoch(
            baseline_model,
            train_sequences_norm,
            baseline_optimizer,
            baseline_criterion,
            device
        )
    )

    validation_loss, validation_accuracy = (
        evaluate_baseline_loss(
            baseline_model,
            validation_sequences_norm,
            baseline_criterion,
            device
        )
    )

    baseline_history.append({
        "epoch": epoch,
        "train_loss": train_loss,
        "validation_loss": validation_loss,
        "train_accuracy": train_accuracy,
        "validation_accuracy": validation_accuracy
    })

    if validation_loss < baseline_best_loss:

        baseline_best_loss = validation_loss
        baseline_epochs_without_improvement = 0

        baseline_best_state = {
            key: value.detach().cpu().clone()
            for key, value in baseline_model.state_dict().items()
        }

        marker = "* BEST"

    else:

        baseline_epochs_without_improvement += 1
        marker = ""

    print(
        f"Epoch {epoch:03d} | "
        f"Train Loss: {train_loss:.4f} | "
        f"Val Loss: {validation_loss:.4f} | "
        f"Train Acc: {train_accuracy:.4f} | "
        f"Val Acc: {validation_accuracy:.4f} "
        f"{marker}"
    )

    if (
        baseline_epochs_without_improvement
        >= BASELINE_PATIENCE
    ):

        print(
            f"\nEarly stopping at epoch {epoch}."
        )

        break

print("\n" + "=" * 70)
print("BASELINE TRAINING COMPLETE")
print("=" * 70)

print(
    "Best validation loss:",
    round(baseline_best_loss, 6)
)

assert baseline_best_state is not None
```
