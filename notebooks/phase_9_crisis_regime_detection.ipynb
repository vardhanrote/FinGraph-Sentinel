```python
market_return = returns_df.mean(axis=1)
```

```python
market_volatility = (
    market_return
    .rolling(20)
    .std()
)
```

```python
market_stress_df = pd.DataFrame({
    "date": market_return.index,
    "market_return": market_return.values,
    "market_volatility": market_volatility.values
})
```

```python
cumulative_market = (
    1 + market_return
).cumprod()
```

```python
running_max = (
    cumulative_market
    .cummax()
)
```

```python
market_drawdown = (
    cumulative_market / running_max
) - 1
```

```python
market_stress_df["drawdown"] = (
    market_drawdown.values
)
```

```python
market_stress_df.tail()
```

```python
cross_sectional_dispersion = (
    returns_df.std(axis=1)
)
```

```python
market_stress_df[
    "return_dispersion"
] = cross_sectional_dispersion.values
```

```python
def rolling_zscore(series, window=60):

    rolling_mean = (
        series
        .rolling(window)
        .mean()
    )

    rolling_std = (
        series
        .rolling(window)
        .std()
    )

    return (
        series - rolling_mean
    ) / rolling_std
```

```python
market_stress_df[
    "volatility_z"
] = rolling_zscore(
    market_stress_df["market_volatility"]
)
```

```python
market_stress_df[
    "dispersion_z"
] = rolling_zscore(
    market_stress_df["return_dispersion"]
)
```

```python
market_stress_df[
    "dispersion_z"
] = rolling_zscore(
    market_stress_df["return_dispersion"]
)
```

```python
market_stress_df[
    "drawdown_stress"
] = -market_stress_df["drawdown"]
```

```python
def classify_stress(value):

    if pd.isna(value):
        return "Unknown"

    if value < 0:
        return "Normal"

    elif value < 1:
        return "Elevated"

    elif value < 2:
        return "High"

    else:
        return "Critical"
```

```python
phase9_df[
    "stress_regime"
] = phase9_df[
    "systemic_stress_index"
].apply(
    classify_stress
)
```

```python
lag_results = []

for lag in range(1, 31):

    lagged_network = (
        phase9_df[
            "network_stress_index"
        ].shift(lag)
    )

    correlation = (
        lagged_network
        .corr(
            phase9_df[
                "market_stress_index"
            ]
        )
    )

    lag_results.append({
        "lag": lag,
        "correlation": correlation
    })
```

```python
lag_df = pd.DataFrame(
    lag_results
)
```

```python
plt.figure(figsize=(12, 6))

plt.plot(
    lag_df["lag"],
    lag_df["correlation"]
)

plt.xlabel("Lag")
plt.ylabel("Correlation")

plt.title(
    "Lagged Network Stress vs Market Stress"
)

plt.tight_layout()

plt.show()
```

```python
phase9_df[
    "early_warning"
] = (
    phase9_df[
        "network_stress_index"
    ] > network_threshold
)
```

```python
future_window = 10
```

```python
phase9_df[
    "future_market_stress"
] = (
    phase9_df[
        "market_stress_index"
    ]
    .rolling(
        future_window,
        min_periods=1
    )
    .max()
    .shift(
        -future_window
    )
)
```

```python
future_threshold = 1.0

phase9_df[
    "future_stress_event"
] = (
    phase9_df[
        "future_market_stress"
    ] >= future_threshold
)
```

```python
phase9_df[
    [
        "date",
        "network_stress_index",
        "early_warning",
        "future_stress_event"
    ]
].tail(20)
```

```python
tp = (
    phase9_df["early_warning"]
    & phase9_df["future_stress_event"]
).sum()

fp = (
    phase9_df["early_warning"]
    & ~phase9_df["future_stress_event"]
).sum()

fn = (
    ~phase9_df["early_warning"]
    & phase9_df["future_stress_event"]
).sum()

tn = (
    ~phase9_df["early_warning"]
    & ~phase9_df["future_stress_event"]
).sum()
```

```python
print("TP:", tp)
print("FP:", fp)
print("FN:", fn)
print("TN:", tn)
```

```python
precision = (
    tp / (tp + fp)
    if (tp + fp) > 0
    else 0
)

recall = (
    tp / (tp + fn)
    if (tp + fn) > 0
    else 0
)

f1 = (
    2 * precision * recall
    / (precision + recall)
    if (precision + recall) > 0
    else 0
)
```

```python
print(
    f"Precision: {precision:.3f}"
)

print(
    f"Recall: {recall:.3f}"
)

print(
    f"F1 Score: {f1:.3f}"
)
```

```python
confusion_matrix = pd.DataFrame(
    [
        [tn, fp],
        [fn, tp]
    ],
    index=[
        "Actual Normal",
        "Actual Stress"
    ],
    columns=[
        "Predicted Normal",
        "Predicted Stress"
    ]
)

confusion_matrix
```
